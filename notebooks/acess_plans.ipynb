```python
import datetime as dt
import polars as pl
```

```python
suscription_plan_schema = {"Plan_Id" : pl.Int64,
                            "Plan_Name" : pl.String,
                            "Plan_Cost" : pl.Float16, 
                            "Plan_Duration" : pl.Int64,
                            "Plan_Start_Time" : pl.Datetime, 
                            "Plan_End_Time" : pl.Datetime, 
                            "Plan_Description" :pl.String,
                            "Updated_At": pl.Datetime
                            }
```

```python
suscriptions = [(1, "Daily_Pass", 150, 1, dt.datetime(2026,1,1), None, "Valid for 24 hours", dt.datetime.now()),
                (2, "Momentum_Pass", 650, 30, dt.datetime(2026,1,1), None, "Valid for 30 days and includes 5 daily passes", dt.datetime.now()),
                (3, "Monthly_Pass", 1200, 30, dt.datetime(2026,1,1), None,  "Valid for 30 days and includes unlimited access", dt.datetime.now()),
                (4, "Yearly_Pass", 10000, 365, dt.datetime(2026,1,1), None,  "Valid for 365 days and includes unlimited access", dt.datetime.now())]

df_suscriptions = pl.DataFrame(suscriptions, schema = suscription_plan_schema, orient = "row").write_csv()
df_suscriptions
```
